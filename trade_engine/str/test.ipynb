```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sol = yf.Ticker('SOL-USD')
df_sol = sol.history(period='1y', interval='1h')
df_sol['normalized'] = df_sol['Close'] / df_sol['Close'].iloc[0]
btc = yf.Ticker('BTC-USD')
df_btc = btc.history(period='1y', interval='1h')
df_btc['normalized'] = df_btc['Close'] / df_btc['Close'].iloc[0]
eth = yf.Ticker('ETH-USD')
df_eth = eth.history(period='1y', interval='1h')
df_eth['normalized'] = df_eth['Close'] / df_eth['Close'].iloc[0]
```

```python
plt.figure(figsize=(14, 7))
plt.plot(df_sol.index, df_sol['normalized'], label='SOL-USD')
plt.plot(df_btc.index, df_btc['normalized'], label='BTC-USD')
plt.plot(df_eth.index, df_eth['normalized'], label='ETH-USD')
plt.title('Price Comparison: SOL, BTC, ETH (1 Year, 1 Hour Interval)')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid()
plt.show()
```
